# Comparison Plots

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
data_k = pd.read_excel('response data.xlsx', sheet_name='k')
data_k_0 = pd.read_excel('response data.xlsx', sheet_name='k=0')
data_k_inf = pd.read_excel('response data.xlsx', sheet_name='k=inf')
data_EI_by_5 = pd.read_excel('response data.xlsx', sheet_name='EI by 5')
data_EI_by_10 = pd.read_excel('response data.xlsx', sheet_name='EI by 10')
data_EI_by_15 = pd.read_excel('response data.xlsx', sheet_name='EI by 15')
# data_EI_by_25 = pd.read_excel('response data.xlsx', sheet_name='EI by 25')
data_kr = pd.read_excel('response data.xlsx', sheet_name='kr')
```

```python
x = np.linspace(0, 12.5, len(data_k['Point ID']))
plt.figure(figsize=(15,5))
plt.plot(x, data_k['abs(elevation)'], label='$k_s$ = 100 N/m')
plt.plot(x, data_k_0['abs(elevation)'], label='$k_s$ = 0')
plt.plot(x, data_k_inf['abs(elevation)'], label='$k_s$ = ∞')
# plt.plot(x, data_k['eta_im'])
# plt.plot(x, data_k['eta_re'])
plt.xlabel('Position along the beam (m)')
plt.ylabel('Absolute elevation amplitude (m)')
# plt.title('Absolute Response Comparison')
plt.grid()
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(15,5))
plt.plot(x, data_k['abs(elevation)'], label='$D_{1} = D_{2}$')
plt.plot(x, data_EI_by_5['abs(elevation)'], label='$D_{1} = D_{2}/5$')
plt.plot(x, data_EI_by_10['abs(elevation)'], label='$D_{1} = D_{2}/10$')
plt.plot(x, data_EI_by_15['abs(elevation)'], label='$D_{1} = D_{2}/15$')
# plt.plot(x, data_EI_by_25['abs(elevation)'], label='$EI_1 = EI_2/25$')
plt.xlabel('Position along the beam (m)')
plt.ylabel('Absolute elevation amplitude (m)')
# plt.title('Absolute Response Comparison')
plt.legend()
plt.grid()
# plt.xlim(0,1)
# plt.ylim(0.004,0.007)
plt.show()
```

```python
fig, axes = plt.subplots(3, 1, figsize=(15, 10), sharex=True)

# Real parts
axes[0].plot(x, data_EI_by_10['eta_re'], label='elastic spring (real)', color='C0')
axes[0].plot(x, data_kr['eta_re'], label='rotational spring (real)', color='C1')
axes[0].set_ylabel('Real(elevation) (m)')
axes[0].legend(loc='upper right')
axes[0].grid(True)

# Imaginary parts
axes[1].plot(x, data_EI_by_10['eta_im'], label='elastic spring (imag)', color='C0')
axes[1].plot(x, data_kr['eta_im'], label='rotational spring (imag)', color='C1')
axes[1].set_ylabel('Imag(elevation) (m)')
axes[1].legend(loc='upper right')
axes[1].grid(True)

# Absolute values
axes[2].plot(x, data_EI_by_10['abs(elevation)'], label='elastic spring (abs)', color='C0')
axes[2].plot(x, data_kr['abs(elevation)'], label='rotational spring (abs)', color='C1')
axes[2].set_xlabel('Position along the beam (m)')
axes[2].set_ylabel('Absolute elevation (m)')
axes[2].legend(loc='upper right')
axes[2].grid(True)

plt.suptitle('Elastic vs Rotational spring: real, imag, abs (top → bottom)')
plt.tight_layout(rect=[0, 0, 1, 0.96])
plt.show()
```

```python
plt.figure(figsize=(15,5))
plt.plot(x, data_EI_by_10['abs(elevation)'], label='elastic spring (abs)', color='C0')
plt.plot(x, data_kr['abs(elevation)'], label='rotational spring (abs)', color='C1')
plt.xlabel('Position along the beam (m)')
plt.ylabel('Absolute elevation (m)')
plt.legend(loc='upper right')
plt.grid(True)
# plt.title('Elastic vs Rotational spring: absolute elevation')
plt.show()
```
